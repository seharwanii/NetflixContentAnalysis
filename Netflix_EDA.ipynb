```python
df["date_added"].head()
```

```python
df["date_added"] = pd.to_datetime(df["date_added"], errors="coerce")
```

```python
df["date_added"].dtype
```

```python
df["date_added"].head()
```

```python
df["added_year"] = df["date_added"].dt.year
```

```python
df["month_added"] = df["date_added"].dt.month
```

```python
df["added_month_name"] = df["date_added"].dt.month_name()
```

```python
df["added_month_name"].head()
```

```python
df["added_year"].value_counts().sort_index()
```

```python
year_added = df["added_year"].value_counts().sort_index()
plt.figure(figsize=(10,6))
sns.lineplot(
    x=year_added.index,
    y=year_added.values,
    marker="o"
)
plt.title("Netflix Titlles added by Year")
plt.xlabel("Year")
plt.ylabel("Number pf Titles")

plt.show()
```

```python
month_count = df["added_month_name"].value_counts()
```

```python
month_order = ["January", "February", "March", "April", "May", "June", "July", "August", "September", "October", "November", "December"]
```

```python
month_count = df["added_month_name"].value_counts().reindex(month_order)
```

```python
plt.figure(figsize=(12,8))
sns.barplot (
    x=month_count.index,
    y=month_count.values
)

plt.title("Netflix Titles added by month")
plt.xlabel("Month")
plt.ylabel("Number of Titles")
plt.xticks(rotation=45)
plt.show()
```

```python
type_year = df.groupby(["added_year","type"]).size().reset_index(name="count")
```

```python
plt.figure(figsize=(12,6))
sns.lineplot(
    data=type_year,
    x="added_year",
    y="count",
    hue="type",
    marker="o",
)

plt.title("Movies vs TV  Shows added over the years")
plt.xlabel("Year")
plt.ylabel("Nummber of Titles")
plt.grid()
plt.show()
```
